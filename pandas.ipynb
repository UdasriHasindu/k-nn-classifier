# Learn Pandas

```python
import pandas as pd
```

```python
data = pd.Series([10, 20, 30, 40, 50])
print(data)
```

```python
# dataframe
dataframe1 = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
    'Age': [24, 27, 22, 32, 29],
    'City': ['New York', 'Los Angeles', 'Chicago', 'Houston', 'Phoenix']
})

print(dataframe1)
```

```python
dataframe2 = pd.DataFrame({
    'Product': ['Laptop', 'Smartphone', 'Tablet', 'Monitor', 'Keyboard'],
    'Price': [1200, 800, 300, 400, 100],
    'Stock': [50, 150, 200, 75, 300]
})

print(dataframe2)
```

```python
# Concatinate DataFrames

df = pd.concat([dataframe1, dataframe2])

df
```

```python
# vertical concatination

df = pd.concat([dataframe1, dataframe2], axis=0, ignore_index=True)
df
```

```python
# Horizontal concatination

df = pd.concat([dataframe1, dataframe2], axis=1, ignore_index=True)
df
```
