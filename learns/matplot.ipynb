```python
df = pd.read_csv('Iris.csv')
df.head()
```

```python
species = df.Species

# plot 1
sepal_length = df.SepalLengthCm

plt.subplot(2,2,1)
plt.bar(species, sepal_length)
plt.ylabel('Sepal Length cm')
plt.title('Sepal Length of Different Species')
plt.xticks(rotation=45)

# plot 2
sepal_width = df.SepalWidthCm

plt.subplot(2,2,2)
plt.bar(species, sepal_width, color='orange')
plt.ylabel('Sepal Width cm')
plt.title('Sepal Width of Different Species')
plt.xticks(rotation=45)

# plot 3
petal_length = df.PetalLengthCm
plt.subplot(2,2,3)
plt.bar(species, petal_length, color='green')
plt.ylabel('Petal Length cm')
plt.title('Petal Length of Different Species')
plt.xticks(rotation=45)

# plot 4
petal_width = df.PetalWidthCm
plt.subplot(2,2,4)
plt.bar(species, petal_width, color='red')
plt.ylabel('Petal Width cm')
plt.title('Petal Width of Different Species')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

Speed compare

```python
# Histogram - NumPy vs Regular Python
import random
import time

COUNT = 1000
# Method 1: Regular Python (your current approach)
start_time = time.time()
num = []
for i in range(COUNT):
    x = random.randint(0, COUNT+1)
    num.append(x)
python_time = time.time() - start_time

# Method 2: NumPy (faster approach)
start_time = time.time()
num_numpy = np.random.randint(0, COUNT, size=COUNT)  # 0 to 999 inclusive
numpy_time = time.time() - start_time

print(f"Python loop time: {python_time:.6f} seconds")
print(f"NumPy time: {numpy_time:.6f} seconds")
print(f"NumPy is {python_time/numpy_time:.2f}x faster")

# Count frequencies using NumPy
unique, counts = np.unique(num_numpy, return_counts=True)

print("\nNumber frequencies (NumPy method):")
for number, count in zip(unique, counts):
    print(f"{number}s - {count}")

# Plot histogram using NumPy data
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.hist(num, bins=20, color='blue', alpha=0.7, edgecolor='black')
plt.xlabel('Numbers')
plt.ylabel('Frequency')
plt.title('Histogram - Python Loop')

plt.subplot(1, 2, 2)
plt.hist(num_numpy, bins=20, color='red', alpha=0.7, edgecolor='black')
plt.xlabel('Numbers')
plt.ylabel('Frequency')
plt.title('Histogram - NumPy')

plt.tight_layout()
plt.show()
```
